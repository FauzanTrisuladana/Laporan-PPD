```python
import pandas as pd
import numpy as np

df_karyawan = pd.DataFrame({
    'Nama': ['Budi', 'Joko', 'Maya', 'Andi', 'Dewi', 'Dian'],
    'Pekerjaan': ['Programer', 'Data Engineer', 'Manager', 'Web Developer', 'Programer', 'Web Developer'],
    'Gaji': [2000, 3000, 2400, 6000, 8000, 6000]
})

print("DataFrame Karyawan:\n", df_karyawan)
agg_gaji_by_pekerjaan = df_karyawan.groupby('Pekerjaan').agg(
    Gaji_Max=('Gaji', 'max'),
    Gaji_Min=('Gaji', 'min'),
    Gaji_Std=('Gaji', 'std')
)

print("\nAgregasi Gaji per Pekerjaan:\n", agg_gaji_by_pekerjaan)
```

```python
import pandas as pd

df_karyawan = pd.DataFrame({
    'Emp_id': ['E001', 'E002', 'E003', 'E004'],
    'Nama': ['Andi', 'Budi', 'Cici', 'Dedi'],
    'Divisi': ['HR', 'IT', 'Marketing', 'IT']
})
print("DataFrame Karyawan:\n", df_karyawan)
df_gaji = pd.DataFrame({
    'ID': ['E001', 'E002', 'E003', 'E005'],
    'Gaji': [50000, 60000, 55000, 70000],
    'Status': ['Aktif', 'Aktif', 'Aktif', 'Tidak Aktif']
})
print("\nDataFrame Gaji:\n", df_gaji)
merged_df = pd.merge(
    df_karyawan,
    df_gaji,
    left_on='Emp_id',
    right_on='ID',
    how='left',
    suffixes=('_karyawan', '_gaji')
)

print("\nDataFrame setelah digabung (Left Join):\n", merged_df)
```
